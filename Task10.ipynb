# Task 10: Data Science Dashboard

```python
# Step 1: Install required libraries
# Run this in the terminal before executing the script: `pip install streamlit pandas matplotlib seaborn plotly`

import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

# Step 2: Load a richer dataset with 10 entries and multiple fields
data = {
    "Movie": ["Movie A", "Movie B", "Movie C", "Movie D", "Movie E", 
              "Movie F", "Movie G", "Movie H", "Movie I", "Movie J"],
    "Rating": [4.5, 2.0, 3.8, 4.2, 1.5, 4.7, 3.0, 4.9, 2.8, 4.6],
    "Sentiment": ["Positive", "Negative", "Neutral", "Positive", "Negative",
                  "Positive", "Neutral", "Positive", "Negative", "Positive"],
    "Genre": ["Drama", "Thriller", "Comedy", "Action", "Horror", 
              "Romance", "Drama", "Sci-Fi", "Fantasy", "Adventure"],
    "Year": [2021, 2020, 2019, 2021, 2022, 2020, 2023, 2018, 2017, 2021],
    "Length (mins)": [120, 95, 110, 130, 100, 140, 150, 125, 105, 135],
    "Box Office (Millions)": [120, 30, 80, 150, 25, 200, 90, 250, 60, 180]
}
df = pd.DataFrame(data)

# Step 3: Configure Streamlit dashboard layout
st.title("Enhanced Data Science Dashboard")
st.header("Interactive Dashboard with Visualizations")

# Step 4: Display the dataset
st.subheader("Dataset")
st.write(df)

# Step 5: Visualize Sentiment Distribution
st.subheader("Sentiment Distribution")
sentiment_counts = df["Sentiment"].value_counts()
fig, ax = plt.subplots()
ax.bar(sentiment_counts.index, sentiment_counts.values, color=['green', 'red', 'blue'])
ax.set_title("Sentiment Counts")
ax.set_xlabel("Sentiments")
ax.set_ylabel("Count")
st.pyplot(fig)

# Step 6: Visualize Ratings by Genre
st.subheader("Ratings by Genre")
fig, ax = plt.subplots()
ax.bar(df["Genre"], df["Rating"], color='orange')
ax.set_title("Ratings by Genre")
ax.set_xlabel("Genres")
ax.set_ylabel("Ratings")
st.pyplot(fig)

# Step 7: Visualize Box Office Revenue
st.subheader("Box Office Revenue by Movie")
fig, ax = plt.subplots()
ax.barh(df["Movie"], df["Box Office (Millions)"], color='purple')
ax.set_title("Box Office Revenue")
ax.set_xlabel("Revenue (in Millions)")
ax.set_ylabel("Movies")
st.pyplot(fig)

# Additional Visualization 1: Heatmap of Correlations
st.subheader("Correlation Heatmap")
fig, ax = plt.subplots()
correlation_matrix = df[["Rating", "Length (mins)", "Box Office (Millions)"]].corr()
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", ax=ax)
ax.set_title("Correlation Between Numeric Features")
st.pyplot(fig)

# Additional Visualization 2: Line Chart for Box Office Over Years
st.subheader("Box Office Revenue Over Years")
box_office_by_year = df.groupby("Year")["Box Office (Millions)"].sum().reset_index()
line_chart = px.line(box_office_by_year, x="Year", y="Box Office (Millions)", 
                     title="Box Office Revenue Trend Over Years")
st.plotly_chart(line_chart)

# Step 8: Add interactivity
selected_movie = st.selectbox("Select a Movie to See Details:", df["Movie"])
movie_details = df[df["Movie"] == selected_movie]
st.write("Details for Selected Movie:")
st.write(movie_details)

# Step 9: Insights and Observations
st.subheader("Insights")
st.write("""
- Positive sentiments dominate among the movies in this dataset.
- Genres like Romance and Sci-Fi have higher ratings on average.
- Box office revenues show strong correlations with ratings.
- Box office revenue trends across years reveal interesting patterns.
""")
```
